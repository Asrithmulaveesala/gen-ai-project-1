```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY2"]=os.getenv("OPENAI_API_KEY2")
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model_name="gpt-4o")
```

```python
response=llm.invoke("who is founder of rundataai uk based company?")
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages([
    ("system","you are an expert in ai and ml"),
    ("human","{input}")
])
prompt
```

```python
chain=prompt | llm
chain.invoke({"input":"who is founder of rundataai uk based company?"})
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()
chain=prompt | llm | output_parser
chain.invoke({"input":"who is founder of rundataai uk based company?"})
```
